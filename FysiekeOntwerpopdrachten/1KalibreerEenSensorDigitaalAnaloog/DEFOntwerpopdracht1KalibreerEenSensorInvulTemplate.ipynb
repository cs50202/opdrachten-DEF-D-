Van Fmin tot Fmax. We gaan Fmin bepalen door eerst heel zacht op de sensor te drukken. ## *Opdracht 6*: Meten en fitten, Leerdoel

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
# Zet hier jouw data in een voor jou handig formaat

massa_fles = 66.4e-3 #kg
# Factor berekenen voor het bepalen van de massa van de fles aangezien de weegschaal het niet meer deed
# We bepalen de massa die wel op de weegschaal zichtbaar is en kijken hoe hoog het water zit. Op basis hiervan heb je een schaal gram/cm
factor = (378.32e-3 - massa_fles)/6.7
#
lengtes = np.array([0, 15.7, 14.85, 13.9, 13.0, 12.0, 10.9, 10.0, 8.9, 8.1, 7.1, 6.0, 4.9, 3.9, 2.8, 2, 1])
massa = (lengtes*factor) + massa_fles

kracht = massa * 9.81

arduino = np.array([120, 930, 920, 910, 890, 880, 860, 845, 820, 790, 740, 690, 660, 560, 360, 200, 140])
u_arduino = np.array([150, 20, 5, 10, 10, 10, 10, 20, 10, 20, 40, 20, 20, 80, 100, 150, 150])
```

```python
def fit(x, a, b, c):
    return b*x**a + c

vals, covs = curve_fit(fit, kracht, arduino)

xtest = np.linspace(np.min(kracht), np.max(kracht), 500)
ytest = fit(xtest, *vals)
```

```python
# Errorbepaling van de kracht
# We bepalen de error in de krachtbepaling aan de hand van de error in de massabepaling. De massa bepalen we door de lengte af te lezen van de maatbeker
# Errorpropagation is als volgt: het verschil in de werkelijke waarde van de massa en de berekende waarde van de massa

m_bepaald = (4.8 * factor)+massa_fles
m_werkelijk = 332.4e-3 #kg

u_m = np.abs(((m_werkelijk - m_bepaald)/m_werkelijk))
u_F = u_m * kracht
```

```python
# Plot hier de meting en fit een lijn door de metingen.
# Maak gebruik van wat je geleerd hebt bij het practicum.

plt.figure()
plt.plot(kracht, arduino, '.', label = "meetwaardes")
plt.errorbar(kracht, arduino, xerr = u_F, yerr = u_arduino, linestyle = 'none', label = 'meetwaardes')
plt.plot(xtest, ytest, '--', label = 'fit')
plt.xlabel("Kracht $(N)$")
plt.ylabel("arduino-waarde $(bits)$")
plt.legend()
plt.savefig("Grafiek", dpi = 450)
plt.show()
```

## *Opdracht 7*: Iteratie nodig?

We hebben geprobeerd om een krachtsensor te ontwikkelen die verschillen in hard en zacht drukken in een elektrische piano kan registreren. 
Onze criteria waren als volgt:
Forte spelen ongveeer 6-7 newton aan druk, zacht spelen ongeveer 1-2 newton aan druk. Dus de sensor moet op 1 decimaal nauwkeurig zijn tussen 0 tot 8 N.

Er zijn genoeg gegevens om een conclusie te trekken over de bruikbaarheid van de gekozen toepassing, aangezien we een duidelijke fit hebben kunnen verkrijgen binnen de waarden van ons doelbereik. Ook is het met de fit mogelijk om de kracht op 1 decimaal nauwkeurig af te lezen op basis van de sensorwaarde. Er is geen iteratie nodig.

## *Opdracht 8*: Conclusie, leerdoel

De krachtsensor is niet bruikbaar voor de gekozen toepassing voor de elektrische piano, aangezien er voornamelijk veel ruis en onzekerheid is bij de lage waarden. Het verschil tussen hard (6-8 N) en zacht (0-3 N) zal de sensor wel kunnen meten, maar de sensor is niet nauwkeurig genoeg om kleinere overgangen te registreren.
